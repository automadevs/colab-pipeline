# 05 - Administração do Kaggle Dataset

Ferramenta operacional independente para construir e publicar uma versão completa de `automamermaid/comfydocs`.

Este fluxo usa Civitai -> staging -> Kaggle Dataset. Não é importado por nenhum runtime Kaggle e não altera o ComfyUI. A publicação só ocorre após preview e confirmação explícita.

```python
from pathlib import Path
import subprocess
import sys

WORKDIR = Path('/content/colab-pipeline')
SCRIPTS_DIR = WORKDIR / 'scripts'
REPO_URL = 'https://github.com/automadevs/colab-pipeline.git'

if WORKDIR.exists():
    subprocess.run(['git', '-C', str(WORKDIR), 'pull', '--ff-only'], check=False)
else:
    subprocess.run(['git', 'clone', '--depth', '1', REPO_URL, str(WORKDIR)], check=True)

if str(SCRIPTS_DIR) not in sys.path:
    sys.path.insert(0, str(SCRIPTS_DIR))

MANAGER = SCRIPTS_DIR / 'kaggle_dataset_manager.py'

if not MANAGER.exists():
    raise FileNotFoundError(
        f'Módulo não encontrado: {MANAGER}'
    )

from kaggle_dataset_manager import run_manager

DATASET = 'automamermaid/comfydocs'
STAGING_DIR = Path('/content/kaggle_dataset_manager')

print('Administração independente do Dataset Kaggle')
print('Digite URLs Civitai; use done para revisar a fila.')
print('Depois use remove <path>, move <old> <new> ou done.')
run_manager(dataset=DATASET, staging_dir=STAGING_DIR)
```
