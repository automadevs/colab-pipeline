```python
# ============================================================
# KAGGLE RUNTIME - 06: SETUP COMFYUI (GITHUB → SSD LOCAL)
# ============================================================
# Este notebook roda no KAGGLE NOTEBOOK (GPU)
# Objetivo: Instalar/atualizar ComfyUI do GitHub no SSD local
# Usa script compartilhado: scripts/comfyui_setup.py
# ============================================================

import sys
from pathlib import Path

# Adicionar scripts ao path
sys.path.insert(0, "/kaggle/working/scripts")

print("=" * 60)
print("SETUP COMFYUI - GITHUB → SSD LOCAL (KAGGLE NOTEBOOK)")
print("=" * 60)

# Configurações
COMFYUI_DIR = Path("/kaggle/working/ComfyUI")
MODELS_DIR = Path("/kaggle/working/ComfyUI/models")
REPO_URL = "https://github.com/comfyanonymous/ComfyUI.git"

# Apenas custom nodes necessários; o Manager é integrado ao ComfyUI.
CUSTOM_NODES = [
    "cubiq/ComfyUI_essentials",
    "lbouaraba/comfyui-krea2edit",
]

print(f"ComfyUI dir: {COMFYUI_DIR}")
print(f"Models dir: {MODELS_DIR}")
print(f"Repo: {REPO_URL}")
print(f"Custom nodes: {CUSTOM_NODES}")

# Executar setup via script compartilhado
from comfyui_setup import setup_comfyui, detect_gpu

try:
    comfyui_path = setup_comfyui(
        comfyui_dir=COMFYUI_DIR,
        repo_url=REPO_URL,
        custom_nodes=CUSTOM_NODES,
        models_dir=MODELS_DIR,
    )
    
    print(f"\n✅ COMFYUI PRONTO em {comfyui_path}")
    print(f"Modelos em {MODELS_DIR}")
    
    # Verificar estrutura de modelos
    for cat in ["checkpoints", "loras", "vae", "controlnet", "upscale_models", "embeddings", "diffusion_models", "text_encoders", "clip", "video_models"]:
        cat_dir = MODELS_DIR / cat
        if cat_dir.exists():
            files = list(cat_dir.iterdir())
            if files:
                print(f"  {cat}: {len(files)} arquivo(s)")
                for f in files[:5]:
                    size_gb = f.stat().st_size / (1024**3)
                    print(f"    - {f.name} ({size_gb:.2f} GB)")
                if len(files) > 5:
                    print(f"    ... e mais {len(files) - 5}")
            else:
                print(f"  {cat}: vazio")
        else:
            print(f"  {cat}: não existe")
            
except Exception as e:
    print(f"\n❌ FALHA: {e}")
    import traceback
    traceback.print_exc()
    raise

print("\n" + "=" * 60)
print("SETUP COMFYUI CONCLUÍDO")
print("=" * 60)
print("\nPRÓXIMO PASSO:")
print("  - Para sync diário: Execute 07_sync_robust.ipynb")
print("  - Para iniciar ComfyUI: Execute 08_master_pipeline.ipynb")
print("  - Ou use scripts/comfyui_setup.py --start --health-check --ngrok")
```
